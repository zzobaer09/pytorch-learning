### Random Tensors

`starts with random tensor -> looks at data -> update random number -> looks at data -> update random number`

```python
# create random tensor of size (3,4)
random_tensor = torch.rand(10,10,10)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.size()
```

```python
random_tensor.numel()
```

##### Randome Number -> Image

```python
random_num_img = torch.rand(size=(3,2,2))
random_num_img
```

```python
plt.imshow(random_num_img.permute(1,2,0))
plt.axis("off")
plt.show()
```
