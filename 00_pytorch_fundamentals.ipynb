##### multiplying tensors

```python
MATRIX_FLOAT_16 = torch.rand(size=(3, 4), dtype=torch.float16,device="cuda")*100
MATRIX_FLOAT_32 = torch.rand(size=(3,4),dtype=torch.float32,device="cuda")*100
```

```python
MATRIX_FLOAT_16,MATRIX_FLOAT_32
```

```python
MATRIX_FLOAT_32 * MATRIX_FLOAT_16
#ERROR IN MATRIX MULTIPLICATION (@) BECAUSE DTYPE DIDNT MATCH
```

```python
MATRIX_M1 = torch.tensor([
        [5,3,-1],
        [-1,2,5],
        [0,0,1]
    ])
MATRIX_M2 = torch.tensor([
        [4,-1],
        [-1,2],
        [3, 4]
    ])
```

```python
MATRIX_M1,MATRIX_M2
```

```python
MATRIX_M1@MATRIX_M2
```
