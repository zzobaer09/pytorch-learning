## Reshaping, Stacking, Squeezing, Unsqueezing tensor
* Reshaping - reshapes input tensor to a defined shape.
* View - return a view of an input tensor of certain shpae but keep the same memory.
* Stacking- Combine multiple tensors on top of eachother(vstack) or side by side(hstack).
* Squeez - removes `1` dimention from a tensor
* Unsqueez - add `1` dimention to a tensor
*Permute - return view of a input tensor with dimention permuted in a certain way.

```python
x = torch.arange(1.,10.)
x,x.shape
```

```python
x_reshaped=x.reshape(9,1)
x_reshaped,x_reshaped.shape
# given a invalid dimention, reshape will throw error. example 1x8 2x5 etc.
```

`View - shares same memory`
changing from `z` will change value of `x`

```python
z = x.view(3,3)
z,z.shape
```

```python
z[0:5] = -99
z,x
```

`stack-on top of eachother`

```python
x_stacked = torch.stack([x,x,x],dim=1)
x_stacked
```

```python
x_stacked = torch.stack([x,x,x],dim=0) # default dim=0
x_stacked
```

`Squeez and Unsqueez`

`Squeez-removes a dimention`

`Unsqeez-adds dimention`

```python
x=torch.tensor([[5,2,1,4,5,1,5,2,2,1,3,1,5]])
```

```python
x_squeezed = x.squeeze()
```

```python
print(f"Before squeezing tensor: {x}")
print(f"Before squeezing shape: {x.shape}")
print(f"After squeezing tensor: {x_squeezed}")
print(f"After squeezing shape: {x_squeezed.shape}")
```

```python
y=torch.tensor([5,2,1,4,5,1,5,2,2,1,3,1,5])
```

```python
y_unsqueezed_dim0 = y.unsqueeze(dim=0)
y_unsqueezed_dim1 = y.unsqueeze(dim=1)
```

`for y unsqueezed at dim 0`

```python
print(f"Before squeezing tensor: {y}")
print(f"Before squeezing shape: {y.shape}")
print(f"After squeezing tensor: {y_unsqueezed_dim0}")
print(f"After squeezing shape: {y_unsqueezed_dim0.shape}")
```

`for y unsqueezed at dim 1`

```python
print(f"Before unsqueezing tensor: {y}")
print(f"Before unsqueezing shape: {y.shape}")
print(f"After unsqueezing tensor (dim=1): {y_unsqueezed_dim1}")
print(f"After unsqueezing shape (dim=1): {y_unsqueezed_dim1.shape}")
```

`Permute-rearranges the dimention of tensor in a specified order`

```python
random_num_img,random_num_img.shape #[color channels,height,width]
```

```python
random_num_img_permuted = torch.permute(random_num_img,(1,2,0)) #[height,width,color channels]
random_num_img_permuted,random_num_img_permuted.shape
```
